```python
import cirq
import numpy as np
import matplotlib.pyplot as plt
```

```python
def apply_phase_estimation(circuit, qubits, phase):
  circuit.append(cirq.H.on_each(*qubits))
  circuit.append(cirq.CZ(qubits[0], qubits[1]) ** phase)
  circuit.append(cirq.H.on_each(*qubits))
  circuit.append(cirq.SWAP(*qubits))
  circuit.append(cirq.measure(*qubits, key='result'))

def simulate_phase_estimation(phase):
  q0, q1 = cirq.LineQubit.range(2)
  circuit = cirq.Circuit()
  apply_phase_estimation(circuit, [q0, q1], phase)
  sim = cirq.Simulator()
  result = sim.run(circuit, repetitions=1000)
  return circuit, result
```

```python
phase = 0.25
circuit, result = simulate_phase_estimation(phase)

print("Circuit:\n", circuit)
print("Result histogram:", result.histogram(key='result'))
```
